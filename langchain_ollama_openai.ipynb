## 1. Connect to Ollama

```python
from langchain_ollama import ChatOllama

ollama_llm = ChatOllama(
    model="llama3.2",                   # must match a name from `ollama list`
    base_url="http://localhost:11434",  # default Ollama address
    temperature=0,
)
P
print(ollama_llm.invoke("Say hello in one short sentence.").content)
```

## 3. Chain: prompt | model | parser

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a concise assistant. Answer in at most two sentences."),
    ("human", "Explain {topic} to a {audience}."),
])

chain = prompt | ollama_llm | StrOutputParser()
# chain = prompt | openai_llm | StrOutputParser()   # OpenAI version

print(chain.invoke({"topic": "Agentic AI", "audience": "beginner"}))
```

## 4. Streaming

```python
for chunk in chain.stream({"topic": "embeddings", "audience": "beginner"}):
    print(chunk, end="", flush=True)
```
